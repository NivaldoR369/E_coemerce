```python
import pandas as pd

# Abrir a aba fCompras
fCompras = pd.read_excel(
    "compras.xlsx",
    sheet_name="fCompras",
    engine="openpyxl"
)

# 1. Quantidade de linhas e colunas
print("===== TAMANHO DA TABELA =====")
print("Quantidade de linhas:", fCompras.shape[0])
print("Quantidade de colunas:", fCompras.shape[1])

# 2. Nomes das colunas
print("\n===== NOMES DAS COLUNAS =====")
print(fCompras.columns.tolist())

# 3. Linhas duplicadas
print("\n===== LINHAS DUPLICADAS =====")
print("Quantidade de linhas duplicadas:", fCompras.duplicated().sum())

# 4. Colunas duplicadas
print("\n===== COLUNAS DUPLICADAS =====")
print("Quantidade de colunas duplicadas:",
      fCompras.columns.duplicated().sum())

# Mostrar os nomes das colunas duplicadas
colunas_duplicadas = fCompras.columns[
    fCompras.columns.duplicated()
]

print("Colunas duplicadas:", colunas_duplicadas.tolist())

# 5. Valores nulos por coluna
print("\n===== VALORES NULOS POR COLUNA =====")
print(fCompras.isnull().sum())

# 6. Total de valores nulos
print("\n===== TOTAL DE VALORES NULOS =====")
print("Total:", fCompras.isnull().sum().sum())

# 7. Quantidade de linhas que possuem algum valor nulo
print("\n===== LINHAS COM VALORES NULOS =====")
print("Quantidade:", fCompras.isnull().any(axis=1).sum())

# 8. Quantidade de colunas que possuem algum valor nulo
print("\n===== COLUNAS COM VALORES NULOS =====")
print("Quantidade:", fCompras.isnull().any(axis=0).sum())
```

```python
import pandas as pd

# 1. Abrir a aba fCompras
fCompras = pd.read_excel(
    "compras.xlsx",
    sheet_name="fCompras",
    engine="openpyxl"
)

# 2. Abrir a aba dfornecedor
dfornecedor = pd.read_excel(
    "compras.xlsx",
    sheet_name="dfornecedor",
    engine="openpyxl"
)

# 3. Transformar as colunas de data em datas
fCompras["Previsão Entrega"] = pd.to_datetime(
    fCompras["Previsão Entrega"],
    errors="coerce"
)

fCompras["Data Entrega"] = pd.to_datetime(
    fCompras["Data Entrega"],
    errors="coerce"
)

# 4. Calcular quantos dias houve de atraso
fCompras["Dias Atraso"] = (
    fCompras["Data Entrega"] - fCompras["Previsão Entrega"]
).dt.days

# 5. Selecionar somente as entregas atrasadas
atrasos = fCompras[fCompras["Dias Atraso"] > 0].copy()

# 6. Relacionar fCompras com dfornecedor
resultado_atrasos = atrasos.merge(
    dfornecedor,
    left_on="Fornecedor",
    right_on="Codigo Fornecedor",
    how="left"
)

# 7. Ranking pela quantidade de entregas atrasadas
ranking_atrasos = (
    resultado_atrasos
    .groupby(["Codigo Fornecedor", "Nome Fornecedor"])
    .size()
    .reset_index(name="Quantidade de Atrasos")
    .sort_values("Quantidade de Atrasos", ascending=False)
)

print("===== FORNECEDORES COM MAIS ENTREGAS ATRASADAS =====")
display(ranking_atrasos)

# 8. Ranking pelo total de dias de atraso
ranking_dias = (
    resultado_atrasos
    .groupby(["Codigo Fornecedor", "Nome Fornecedor"])["Dias Atraso"]
    .agg(
        Total_Dias_Atraso="sum",
        Media_Dias_Atraso="mean",
        Maior_Atraso="max"
    )
    .sort_values("Total_Dias_Atraso", ascending=False)
)

print("===== FORNECEDORES COM MAIOR TOTAL DE DIAS DE ATRASO =====")
display(ranking_dias)
```

```python
import pandas as pd

# ==========================================
# 1. ABRIR AS ABAS DO ARQUIVO
# ==========================================

fCompras = pd.read_excel(
    "compras.xlsx",
    sheet_name="fCompras",
    engine="openpyxl"
)

dFornecedor = pd.read_excel(
    "compras.xlsx",
    sheet_name="dFornecedor",
    engine="openpyxl"
)

# ==========================================
# 2. CONVERTER AS DATAS
# ==========================================

fCompras["Previsão Entrega"] = pd.to_datetime(
    fCompras["Previsão Entrega"],
    errors="coerce"
)

fCompras["Data Entrega"] = pd.to_datetime(
    fCompras["Data Entrega"],
    errors="coerce"
)

# ==========================================
# 3. CALCULAR OS DIAS DE ATRASO
# ==========================================

fCompras["Dias Atraso"] = (
    fCompras["Data Entrega"] -
    fCompras["Previsão Entrega"]
).dt.days

# ==========================================
# 4. SELECIONAR APENAS ENTREGAS ATRASADAS
# ==========================================

atrasos = fCompras[
    fCompras["Dias Atraso"] > 0
].copy()

print("Total de entregas atrasadas:", len(atrasos))

# ==========================================
# 5. RELACIONAR COM A TABELA DE FORNECEDORES
# ==========================================

resultado_atrasos = atrasos.merge(
    dFornecedor,
    left_on="Fornecedor",
    right_on="Codigo Fornecedor",
    how="left"
)

# ==========================================
# 6. VERIFICAR O RESULTADO DO CRUZAMENTO
# ==========================================

print("\nColunas da tabela resultante:")
print(resultado_atrasos.columns.tolist())

# ==========================================
# 7. RANKING DE FORNECEDORES
# ==========================================

ranking_atrasos = (
    resultado_atrasos
    .groupby("Nome Fornecedor")
    .size()
    .sort_values(ascending=False)
)

print("\n===== FORNECEDORES QUE MAIS ATRASARAM =====")
display(ranking_atrasos)
```
